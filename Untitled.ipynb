```python
import sys, os
sys.path.insert(0, os.path.abspath("src"))
from dotenv import load_dotenv
load_dotenv(override=True)

import importlib, drafting, verification
importlib.reload(drafting); importlib.reload(verification)
from collections import defaultdict, Counter
from drafting import draft, SYSTEM_C, outcome
from verification import verify
from db import get_connection
from retrieval import DenseChunks
from llm import LLM

conn = get_connection()
chunked = DenseChunks(conn)
gem = LLM(min_interval=4.5)                                  # gemini-3.5-flash-lite, from .env
gem_v = LLM(model="gemini-3.5-flash", min_interval=6.5)      # a different model checks the drafts

gold = defaultdict(set)
for r in conn.execute("SELECT intent, doc_id FROM intent_docs"):
    gold[r["intent"]].add(r["doc_id"])

dev_all = conn.execute("SELECT ticket_id, intent, query FROM tickets WHERE split='dev'").fetchall()
dev10 = dev_all[:10]


def run_drafts(rows, system=SYSTEM_C):
    return [(r, draft(gem, chunked, r["query"], system=system)) for r in rows]


res_dev_c = run_drafts(dev_all)
print("model:", gem.model, "| real calls:", gem.calls, "| cache hits:", gem.cache_hits)
print(Counter(outcome(o) for _, o in res_dev_c))
```

```python
import json
from pathlib import Path
import pandas as pd
from verification import split_sentences, verify

records = json.loads(Path("results/verification_drafts.json").read_text(encoding="utf-8"))
print("drafts loaded:", len(records))

# A. export the labelling sheets first (no API calls)
sent_rows, draft_rows, view = [], [], []
for rec in records:
    sents = split_sentences(rec["draft"])
    for i, s in enumerate(sents, 1):
        sent_rows.append({"draft_id": rec["draft_id"], "sentence_id": i, "sentence": s, "label": ""})
    draft_rows.append({"draft_id": rec["draft_id"], "query": rec["query"],
                       "draft": rec["draft"], "answers_question": ""})
    view.append(f"## Draft {rec['draft_id']}\n**Customer:** {rec['query']}\n\n**Sentences:**\n" +
                "\n".join(f"{i}. {s}" for i, s in enumerate(sents, 1)) + "\n\n**Passages shown to the model:**\n" +
                "\n".join(f"[{i}] ({d}) {t}" for i, (d, t, _s) in enumerate(rec["passages"], 1)) + "\n\n---\n")
pd.DataFrame(sent_rows).to_csv("results/labels_sentences.csv", index=False, encoding="utf-8-sig")
pd.DataFrame(draft_rows).to_csv("results/labels_drafts.csv", index=False, encoding="utf-8-sig")
Path("results/labelling_view.md").write_text("\n".join(view), encoding="utf-8")
print("sentences to label:", len(sent_rows))

# B. run the verifier, but stop early if the service is still down
outputs, failed, in_a_row = {}, [], 0
for rec in records:
    draft_out = {"draft": rec["draft"], "abstained": False, "cited": rec["cited"],
                 "passages": [tuple(p) for p in rec["passages"]]}
    try:
        outputs[rec["draft_id"]] = verify(gem_v, draft_out, rec["query"])
        in_a_row = 0
    except Exception as e:
        failed.append(rec["draft_id"]); in_a_row += 1
        print("draft", rec["draft_id"], "failed:", getattr(e, "code", type(e).__name__))
        if in_a_row >= 2:
            print("two failures in a row: stopping, try again later")
            break
Path("results/verifier_outputs.json").write_text(json.dumps(outputs, indent=1), encoding="utf-8")
print("verified:", len(outputs), "| not done:", len(records) - len(outputs))
```
